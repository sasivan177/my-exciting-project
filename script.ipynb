## Task 1: Create a Git repository

Create a Git repository named `my-exciting-project`.

````bash
mkdir my-exciting-project && cd my-exciting-project
git init
git commit --allow-empty -m "Initial commit" 
````

## Task 2: Create standard GitFlow branches

Create a `develop` branch from the default `master` branch.

````bash
git branch develop
````

## Task 3: Create and commit the initial feature branch

Switch to `develop`, create `feature/initial-setup`, and add these files:

- `README` containing `This is a sample README`
- `LICENSE` containing `This is a sample LICENSE`
- `my-awesome-script.py` containing:

  ```python
  #!/bin/python
  print('Hello, World!')
  ```

Stage and commit all changes on the current feature branch.

````bash
git checkout develop
git checkout -b feature/initial-setup
````

````bash

echo "This is a sample README" > README
echo "This is a sample LICENSE" > LICENSE
cat > my-awesome-script.py << 'EOF'
#!/bin/python
print('Hello, World!')
EOF
````

````bash
git add README LICENSE my-awesome-script.py
git commit -m "Add initial README, LICENSE, and script"
````

## Task 4: Merge feature changes into the main branches

Merge `feature/initial-setup` into `develop`, then merge `develop` into `master`.

````bash
git checkout develop
git merge feature/initial-setup -m "Merge feature/initial-setup into develop"
````

````bash

git checkout master
git merge develop -m "Merge develop into master"
````

3. Merge `feature/enhancement-1` into `develop` first.
4. Merge `feature/enhancement-2` into `develop`; resolve the resulting conflict.

````bash
git checkout develop
git checkout -b feature/enhancement-1
````

````bash
git add my-awesome-script.py
git commit -m "Change greeting to Howdy, World!"
````

````bash
git merge feature/my-enhacement-2 -m "Merge feature/my-enhacement-2 into develop"
````

````bash
git add my-awesome-script.py
git commit -m "Merge feature/my-enhacement-2 into develop (resolve conflict)"
````

## Task 6: Amend a commit

From `develop`, create `feature/next-awesome-thing`. Add the comment `# This is an awesome Python script` on the line below the shebang. Commit with the message `Added a comment`, then amend the commit message to `Added an awesome comment`.

````bash
git checkout develop
git checkout -b feature/next-awesome-thing
````

````bash
sed -i '1a # This is an awesome Python script' my-awesome-script.py
````

````bash
git add my-awesome-script.py
git commit -m "Added a comment"
git commit --amend -m "Added an awesome comment"
````

## Task 7: Drop a commit

While still on `feature/next-awesome-thing`, drop the last commit.

````bash
git reset --hard HEAD~1
````

## Task 8: Squash commits

From `develop`, create `feature/part-1-finale`. Make three separate commits, changing the print statement in sequence to:

1. `Hello! World, how is it going?`
2. `Hello! World, hope you're doing well!`
3. `Hello! World, we are one!`

Squash the last three commits into one commit.

````bash
git checkout develop
git checkout -b feature/part-1-finale
````

````bash
sed -i "s/print('.*')/print('Hello! World, how is it going?')/" my-awesome-script.py
git add my-awesome-script.py
git commit -m "Update greeting v1"

sed -i "s/print('.*')/print(\"Hello! World, hope you're doing well!\")/" my-awesome-script.py
git add my-awesome-script.py
git commit -m "Update greeting v2"

sed -i "s/print('.*')/print('Hello! World, we are one!')/" my-awesome-script.py
git add my-awesome-script.py
git commit -m "Update greeting v3"
````

````bash
git rebase -i HEAD~3
````
